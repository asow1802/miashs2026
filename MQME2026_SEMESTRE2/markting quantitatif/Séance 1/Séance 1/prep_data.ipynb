```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime

sns.set()
```

# I/ Importation des données

### I.A/ Fichiers uniques

```python
# Import des données
df_sales = pd.read_csv("DATA/df_sales.csv")
df_events = pd.read_csv("DATA/df_events.csv")
df_customers = pd.read_csv("DATA/df_customers.csv")

# Suppression des colonnes inutiles
df_sales = df_sales[df_sales.columns.drop(['Unnamed: 0'])]
df_customers=df_customers[df_customers.columns.drop(['Unnamed: 0'])]
df_events=df_events[df_events.columns.drop(['Unnamed: 0'])]

# Conversion des types
df_sales['id_client']=df_sales['id_client'].astype(str)
df_customers['ID']=df_customers['ID'].astype(str)

df_sales.head()
```

### I.B/ Jointure des données

```python
# Jointure des données
df_temp = pd.merge(left=df_sales, right=df_events, left_on='source', right_on='Source')
df_temp = pd.merge(left=df_temp, right=df_customers, left_on='id_client', right_on='ID')
print(f"Nombre de lignes : {df_temp.shape[0]}")
df_temp.head()
```

# II/ Préparation des données

### II.A/ Conversion des types

```python
# Suppression des colonnes superflues
drop_columns = ['ID_x', 'ID_y']
df_work = df_temp[df_temp.columns.drop(drop_columns)]

# Changement des types
df_work['date_creation'] = pd.to_datetime(df_work['date_creation'])
df_work['date_utilisation'] = pd.to_datetime(df_work['date_utilisation'])
df_work['begin_on'] = pd.to_datetime(df_work['begin_on'])
df_work['DDN'] = pd.to_datetime(df_work['DDN'])

df_work.head()
```

```python
# Affichage des types 
df_work.dtypes
```

### II.B/ Traitement des valeurs nulles

```python
# Affichage des taux de remplissage des 
df_work.notnull().mean()
```

```python
### Date utilisation
# Calcule des médianes groupées
df_work['date_utilisation'] = df_work['date_utilisation'].fillna(
    df_work.groupby('source')['date_utilisation']
           .transform(lambda x: x.dropna().median())
)

### Sexe
df_work['sexe']=df_work['sexe'].replace({'1':'F', '0':'M'}) # Formatage des données pour n'avoir que 2 modalités
df_work['sexe']=df_work['sexe'].fillna(value=df_work['sexe'].mode()[0])

### DDN
df_work['DDN']=df_work['DDN'].fillna(df_work['DDN'].median())

### sub, sub_notif, sub_newsletter
df_work['sub']=df_work['sub'].fillna(0)
df_work['sub_notif']=df_work['sub_notif'].fillna(0)
df_work['sub_newsletter']=df_work['sub_newsletter'].fillna(0)
```

### II.C/ Création de variables

```python
# Âge du client
df_work['age_event'] = (df_work['date_creation'] - df_work['DDN']).dt.days // 365

# Anticipation
df_work['anticipation']=(df_work['begin_on'] - df_work['date_creation']).dt.days

# Année et mois de l'évènement
df_work['begin_on_year'] = df_work['begin_on'].dt.year
df_work['begin_on_month'] = df_work['begin_on'].dt.month
df_work['JDS'] = df_work['date_creation'].dt.dayofweek
```

# III/ Table client

### III.A/ Création de la table

##### III.A.1/ Calcul groupés intermédiaires

```python
# Calcul des montants clients
df_clients = (
    df_work
        .groupby('id_client')
        .agg(
            DDN_max=('DDN', 'max'),
            sexe_mode=('sexe', lambda x: x.mode().iloc[0]),
            prix_total=('prix', 'sum'),
            nb_billets=('id_billet', 'nunique'),
            nb_commandes=('commande', 'nunique'),
            nb_sources=('source', 'nunique'),
            anticipation_moy=('anticipation', 'mean'),
            date_creation_min=('date_creation', 'min'),
            date_creation_max=('date_creation', 'max')
        )
        .reset_index()
)

# Calcul de l'âge des clients
df_clients['age'] = (datetime.datetime.now() - df_clients['DDN_max']).dt.days // 365

# Création de colonnes calculées
df_clients['prix_billets_moy'] = df_clients['prix_total'] / df_clients['nb_billets']
df_clients['billets_par_commandes'] = df_clients['nb_billets'] / df_clients['nb_commandes']
df_clients['DDV'] = (df_clients['date_creation_max'] - df_clients['date_creation_min']).dt.days
df_clients['DDV'] = np.maximum(df_clients['DDV'], 1)
df_clients['DDV_annee'] = df_clients['DDV'] / 365
df_clients['DDV_mois'] = df_clients['DDV_annee'] * 12
df_clients['recence'] = (datetime.datetime.now() - df_clients['date_creation_max']).dt.days

### Calcul du score RFM
# Calcul des scores
df_clients['R_score'] = pd.qcut(
    df_clients['recence'],
    q=5,
    labels=range(5, 0, -1)
).astype(str)

df_clients['F_score'] = pd.qcut(
    df_clients['nb_commandes'].rank(method='first'),
    q=5,
    labels=range(1, 6)
).astype(str)

df_clients['M_score'] = pd.qcut(
    df_clients['prix_total'],
    q=5,
    labels=range(1, 6)
).astype(str)

# Création du score concaténé
df_clients['RFM'] = df_clients['R_score'] + df_clients['F_score'] + df_clients['M_score']

df_clients.head()
```

##### III.A.2/ Calcul groupés avancés


- 1 table par décomposition tel que le produit des colonnes nous donne le CA
- niv 1 : nb_clients x panier_moyen
- niv 2 : nb_clients x nb_billets x pm_billets
- niv 3 : nb_clients x CA_moyen_periode x DDV
- Décomposer selon plusieurs critères : Sexe, TA, Anticipation, RFM

```python
### Création de varibale qualitative à ventiler
# Âge
bins = [0, 20, 23, 26, np.inf]
labels = [
    "-20 ans",
    "20-22 ans",
    "23-25 ans",
    "+25 ans"
]

df_clients['age_categorie'] = pd.cut(
    df_clients['age'],
    bins=bins,
    labels=labels,
    right=False  # bornes inclusives à gauche
)

# Anticipation
bins = [-np.inf, 1, 6, 13,np.inf]
labels = [
    'J-J',
    'S1', 'S2', '+S2'
]

df_clients['anticipation_categorie'] = pd.cut(
    df_clients['anticipation_moy'],
    bins=bins,
    labels=labels,
    right=False  # bornes inclusives à gauche
)

# RFM
# TBC : Très bon client
# CF : Client Fidèle
# CR : Client Récens
# CPP : Client à ne pas perdre
# CEV :  Client en veille
# CP : Client Perdus
# NA : Non caractérisé

rfm_dict = {
    # TBC
    '555': 'TBC',

    # CF
    '554': 'CF', '545': 'CF', '455': 'CF',
    '544': 'CF', '445': 'CF', '454': 'CF', '444': 'CF',

    # CR
    '511': 'CR', '512': 'CR', '513': 'CR', '514': 'CR', '515': 'CR',
    '521': 'CR', '522': 'CR', '523': 'CR', '524': 'CR', '525': 'CR',
    '531': 'CR', '532': 'CR', '534': 'CR', '535': 'CR',

    '411': 'CR', '412': 'CR', '413': 'CR', '414': 'CR', '415': 'CR',
    '421': 'CR', '422': 'CR', '423': 'CR', '424': 'CR', '425': 'CR',
    '431': 'CR', '432': 'CR', '433': 'CR', '434': 'CR', '435': 'CR',

    # CPP
    '355': 'CPP', '345': 'CPP', '354': 'CPP', '344': 'CPP',

    # CEV
    '333': 'CEV', '332': 'CEV', '331': 'CEV',
    '323': 'CEV', '322': 'CEV', '321': 'CEV',
    '313': 'CEV', '312': 'CEV', '311': 'CEV',

    # CP
    '111': 'CP', '112': 'CP', '113': 'CP',
    '121': 'CP', '122': 'CP', '123': 'CP',
    '131': 'CP', '132': 'CP', '133': 'CP',
    '211': 'CP', '212': 'CP', '213': 'CP',
    '221': 'CP', '222': 'CP', '223': 'CP',
    '231': 'CP', '232': 'CP', '233': 'CP'
}
df_clients['RFM_cat'] = df_clients['RFM'].map(rfm_dict).fillna('NA')
ordre_RFM=['TBC', 'CF', 'CR', 'CPP', 'CEV', 'CP', 'NA']
```
