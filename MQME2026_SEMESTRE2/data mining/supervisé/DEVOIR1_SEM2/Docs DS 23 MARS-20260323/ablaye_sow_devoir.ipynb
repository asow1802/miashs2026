## 8. Sélection de variables basée sur l'AUC

Le modèle précédent nous révèle que les variables Facebook, TropFacebook et Ordi sont significativement associées à la variable cible "TropInternet". Ce pendant nous allons effectuer une sélection de variables parmi celles restantes pour essayer d'améliorer les performances du modèle. 

```python
best_candidats = [["Facebook", "TropFacebook", "Ordi"], 
                  ["Facebook", "TropFacebook", "Ordi", "Douche"], 
                  ["Facebook", "TropFacebook", "Ordi", "Habillement"],
                  ["Facebook", "TropFacebook", "Ordi", "Douche", "Habillement"]]


auc_scores = []
y = dataDS["TropInternet"]
for candidats in best_candidats:
    logit_model = sm.Logit(y, sm.add_constant(dataDS[candidats]))
    logit_results = logit_model.fit()
    auc = roc_auc_score(y, logit_results.predict(sm.add_constant(dataDS[candidats])))
    auc_scores.append(auc)

for i, candidats in enumerate(best_candidats):
    print(f"Variables: {candidats}, AUC: {auc_scores[i]:.4f}")
```

Ainsi nous constatons que le modèle avec ['Facebook', 'TropFacebook', 'Ordi', 'Habillement'] est celui qui a la meilleure performance avec une AUC de 0.9548.
