```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[2, 12],
              [4, 10],
              [5, 7],
              [4, 5]]) 

def barycentre(*points):
    points_array = np.array(points)
    return np.mean(points_array, axis=0)
g1 = barycentre(X[0], X[1], X[2])
g2 = barycentre(X[3])
print("g1:", g1)
print("g2:", g2)
    
```

```python
plt.scatter(X[:, 0], X[:, 1], color='blue', label='Data Points')
plt.scatter(g1[0], g1[1], color='red', marker='o', s=100, label='g1')
plt.scatter(g2[0], g2[1], color='green', marker='o', s=100, label='g2')
plt.xlabel('absicisses X1')
plt.ylabel('ordonnées X2')
plt.title('K-means Clustering')
plt.legend()
plt.grid()
plt.show()
```

```python
def distance(point1, point2):
    return np.linalg.norm(point1 - point2)
for i, x in enumerate(X):
    d1 = distance(x, g1)
    d2 = distance(x, g2)
    print(f"Distance entre point {i+1} et g1: {d1:.2f}, et g2: {d2:.2f}")
```

x0 et x1 vont avec g1 : c12 = {x1, x2}
x2 et x3 vont avec g2 : c22 = {x3, x4}

```python
new_g1 = barycentre(X[0], X[1])
new_g2 = barycentre(X[2], X[3])
print("new g1:", new_g1)
print("new g2:", new_g2)
```

```python
for i, x in enumerate(X):
    d1 = distance(x, new_g1)
    d2 = distance(x, new_g2)
    print(f"Distance entre point {i+1} et new g1: {d1:.2f}, et new g2: {d2:.2f}")
```
