```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
```

```python
XA = np.linspace(0, 500, 1000)
XB = np.linspace(0, 500, 1000)

YA = []
YB = []
for i in range (1000):
    if 4*XA[i] + 2*XB[i] <= 320 and 4*XA[i] + 10*XB[i] <= 800:
        YA.append (XA[i])
        YB.append (XB[i])
        
X_grid, Y_grid = np.meshgrid (YA, YB)
Z_grid = X_grid*50 + Y_grid*75
```

```python
# Plot
from matplotlib import cm
import matplotlib.pyplot as plt

from matplotlib import cm


fig = plt.figure(figsize=(6, 20))
ax = fig.add_subplot(111, projection='3d')

ax.set_xlabel('YA')
ax.set_ylabel('YB')
ax.set_zlabel('Z')

surf = ax.plot_surface(
    X_grid, Y_grid, Z_grid,
    cmap=cm.winter, alpha=1
)
Z_grid_zero = np.zeros_like(Z_grid)
ax.contourf(X_grid, Y_grid, Z_grid_zero, zdir='z', offset=0, cmap=cm.winter, alpha=0.5)

print(Z_grid_zero)
surf = ax.plot_surface(
    X_grid, Y_grid, Z_grid_zero,
    cmap=cm.winter, alpha=0.5, edgecolor='none'
)

plt.show()
```
