```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KernelDensity
from sklearn.metrics.pairwise import rbf_kernel
```

```python
# Génération données
np.random.seed(0)
X = np.linspace(0, 10, 100)
y = np.sin(X) + np.random.normal(0, 0.3, 100)

def kernel_regression(x0, X, y, h):
    weights = np.exp(-(X - x0)**2 / (2*h**2))
    return np.sum(weights * y) / np.sum(weights)

X_test = np.linspace(0, 10, 200)
y_pred = np.array([kernel_regression(x0, X, y, h=0.5) for x0 in X_test])

plt.scatter(X, y)
plt.plot(X_test, y_pred, color='red')
plt.show()
```

```python
import statsmodels.api as sm

lowess = sm.nonparametric.lowess
z = lowess(y, X, frac=0.2)

plt.scatter(X, y)
plt.plot(z[:,0], z[:,1], color='red')
plt.show()
```
