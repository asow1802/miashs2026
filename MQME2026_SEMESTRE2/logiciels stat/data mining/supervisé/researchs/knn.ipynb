```python
import numpy as np
import matplotlib.pyplot as plt
# import knn pour segmenter les données
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import make_moons
```

```python
data = make_moons(n_samples=200, noise=0.1, random_state=42)
X = data[0]
y = data[1]
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.show()
```

```python
knnModel = KNeighborsClassifier(n_neighbors=5)
knnModel.fit(X, y)
predictions = knnModel.predict(X)
score = knnModel.score(X, y)
print("Accuracy:", score)
plt.scatter(X[:, 0], X[:, 1], c=predictions)
plt.show()
```

```python
inertie = []
k_range = range(1, 10)
fig, ax = plt.subplots(1,10, figsize=(20,5))
col = ["red","yellow","blue","green","purple","orange","pink","brown","gray","cyan"]
for k in k_range:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X, y)
    inertie.append(knn.score(X, y))
    ax[k-1].scatter(X[:, 0], X[:, 1], c=[col[i] for i in knn.predict(X)])
    ax[k-1].set_title(f'k={k}')
    ax[k-1].axis('off')
plt.plot(k_range, inertie)
plt.xlabel('Nombre de voisins k')
plt.ylabel("Inertie (accuracy)")
plt.show()

x = np.array([[0.5, 0.5]])
pred = knnModel.predict(x)
print(f'La prédiction pour le point {x} est la classe {pred}')
```
