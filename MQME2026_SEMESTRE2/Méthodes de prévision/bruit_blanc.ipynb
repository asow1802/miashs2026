```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def bruit_blanc(n, mu=0, sigma=1):
    """
    Génère un signal de bruit blanc gaussien.

    Parameters:
    n (int): Nombre d'échantillons à générer.
    mu (float): Moyenne du bruit (par défaut 0).
    sigma (float): Écart-type du bruit (par défaut 1).

    Returns:
    np.ndarray: Un tableau de bruit blanc gaussien.
    """
    return np.random.normal(mu, sigma, n)
# Exemple d'utilisation
n_samples = 1000
bruit = bruit_blanc(n_samples)
# Affichage du signal de bruit blanc
plt.figure(figsize=(10, 4))
plt.plot(bruit, label='Bruit Blanc Gaussien')
plt.title('Signal de Bruit Blanc Gaussien')
plt.xlabel('Échantillons')
plt.ylabel('Amplitude')
plt.legend()
plt.grid()
plt.show()
```

```python
def autocorrelation(signal):
    """
    Calcule l'autocorrélation d'un signal.

    Parameters:
    signal (np.ndarray): Le signal pour lequel calculer l'autocorrélation.

    Returns:
    np.ndarray: L'autocorrélation du signal.
    """
    n = len(signal)
    autocorr = np.correlate(signal, signal, mode='full')
    return autocorr[n-1:]
# Calcul de l'autocorrélation du signal de bruit blanc
autocorr_bruit = autocorrelation(bruit)
# Affichage de l'autocorrélation
plt.figure(figsize=(10, 4))
plt.plot(autocorr_bruit, label='Autocorrélation du Bruit Blanc')
plt.title('Autocorrélation du Signal de Bruit Blanc')
plt.xlabel('Décalage')  
plt.ylabel('Autocorrélation')
plt.legend()
plt.grid()
plt.show()
```
