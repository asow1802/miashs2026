```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
serbia = [12.7, 14.3, 14.5, 14.3, 15.9, 21.0, 19.8, 19.1, 20.8, 21.1, 22.9, 22.0, 21.1, 20.3, 20.3, 21.4, 26.3]
years = list(range(2004, 2021))

data = pd.DataFrame({'Year': years, 'Serbia': serbia})
#data.to_csv('serbia.csv', index=False)
#data.to_excel('serbia.xlsx', index=False)

data.head()
plt.plot(data['Year'], data['Serbia'], marker='o')
plt.title('Pourcentage d\'énergie renouvelable en Serbie (2004-2020)')
plt.xlabel('Année')
plt.ylabel('Renewable Energy (%)')
plt.grid()
plt.show()
```
